```python

# Reload original CSV so we can inspect untouched date values
from pathlib import Path
import pandas as pd

DATA_PATH = Path(
    "data/raw/Supermart Grocery Sales - Retail Analytics Dataset.csv"
)

raw_df = pd.read_csv(DATA_PATH)

print("Total rows:", len(raw_df))

print("\n--- Original Order Date samples ---")
display(raw_df[["Order Date"]].head(15))

print("\n--- Original date data type ---")
print(raw_df["Order Date"].dtype)

print("\n--- Original date strings ---")
for value in raw_df["Order Date"].dropna().unique()[:15]:
    print(repr(value))

print("\n--- State values ---")
print(raw_df["State"].value_counts(dropna=False))
```

```python

# Parse each date format separately
dates = raw_df["Order Date"].astype("string").str.strip()

hyphen_mask = dates.str.contains("-", na=False)
slash_mask = dates.str.contains("/", na=False)

parsed_dates = pd.Series(
    pd.NaT, index=raw_df.index, dtype="datetime64[ns]"
)

# Format 1: DD-MM-YYYY, e.g. 11-08-2017
parsed_dates.loc[hyphen_mask] = pd.to_datetime(
    dates.loc[hyphen_mask],
    format="%d-%m-%Y",
    errors="coerce"
)

# Format 2: M/D/YYYY, e.g. 4/15/2018
parsed_dates.loc[slash_mask] = pd.to_datetime(
    dates.loc[slash_mask],
    format="mixed",
    errors="coerce"
)

print("Total rows:", len(raw_df))
print("Successfully parsed:", parsed_dates.notna().sum())
print("Invalid or unparsed:", parsed_dates.isna().sum())

print("\nDate range:")
print("Start:", parsed_dates.min())
print("End:", parsed_dates.max())

print("\nSample verification:")
display(pd.DataFrame({
    "original_date": raw_df["Order Date"],
    "parsed_date": parsed_dates
}).head(15))
```

```python
from pathlib import Path
import pandas as pd

clean_df = raw_df.copy()
```

```python
clean_df.columns = clean_df.columns.str.strip()
```

```python
clean_df["Order Date"] = parsed_dates
```

```python
text_cols = clean_df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_cols:
    clean_df[col] = clean_df[col].str.strip()
```

```python
numeric_cols = ["Sales", "Discount", "Profit"]

for col in numeric_cols:
    clean_df[col] = pd.to_numeric(
        clean_df[col],
        errors="coerce"
    )
```

```python
print("Original rows:", len(raw_df))
print("Cleaned rows:", len(clean_df))
```

```python
display(clean_df.isna().sum().to_frame("missing_count"))
```

```python
display(clean_df["State"].value_counts(dropna=False))
```

```python
assert len(clean_df) == len(raw_df), "Unexpected row count change!"

assert clean_df["Order Date"].notna().all(), "Invalid dates remain!"

assert clean_df["Sales"].notna().all(), "Invalid sales values!"

assert clean_df["Profit"].notna().all(), "Invalid profit values!"

print("\nAll validation assertions passed.")
```

```python
from pathlib import Path

output_path = Path("data/processed")
output_path.mkdir(parents=True, exist_ok=True)

csv_path = output_path / "supermart_clean.csv"

clean_df.to_csv(
    csv_path,
    index=False,
    date_format="%Y-%m-%d"
)

print("Saved to:", csv_path)
print("Rows exported:", len(clean_df))
print("File size (KB):", round(csv_path.stat().st_size / 1024, 2))
```
